```python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer
from peft import PeftModel

BASE_MODEL = "mistralai/Mistral-7B-Instruct-v0.3"
LORA_MODEL = "hane123/legal-mistral-7b"

print("⏳ Loading tokenizer...")
tokenizer = AutoTokenizer.from_pretrained(BASE_MODEL)
tokenizer.pad_token = tokenizer.eos_token

print("⏳ Loading base Mistral-7B (~5 minutes)...")
base_model = AutoModelForCausalLM.from_pretrained(
    BASE_MODEL,
    torch_dtype=torch.float16,
    device_map={"": "cpu"}  # force everything to CPU
)

print("⏳ Loading LoRA adapters...")
model = PeftModel.from_pretrained(
    base_model, 
    LORA_MODEL,
    device_map={"": "cpu"}  # force to CPU
)

print("⏳ Merging LoRA into base model...")
model = model.merge_and_unload()
model.eval()

print("✅ Merge complete!")
```

```python
MERGED_MODEL_NAME = "hane123/legal-mistral-7b-merged"

print(f"⏳ Uploading merged model to HuggingFace...")
print(f"📍 Destination: {MERGED_MODEL_NAME}")
print(f"⚠️ This will take ~20-30 minutes (14GB)!")
print(f"☕ Perfect time for a long break!\n")

model.push_to_hub(MERGED_MODEL_NAME)
tokenizer.push_to_hub(MERGED_MODEL_NAME)

print(f"\n✅ Done!")
print(f"🌍 Model available at:")
print(f"   https://huggingface.co/{MERGED_MODEL_NAME}")
```
